```python
import random                  # random 모듈 추가

player_name = "뉴비용사"     # 캐릭터 이름
player_hp = 250                 # 체력
gold = 1500               # 가진 돈
monster_name1 = "황금고블린"  #적 몬스터 이름
monster_name2 = "슬라임"  #적 몬스터 이름
monster_name3 = "오크"  #적 몬스터 이름
monster_hp1 = 110          #적 몬스터 체력
monster_hp2 = 110          #적 몬스터 체력
monster_hp3 = 110          #적 몬스터 체력
monster_gold1 = 3500
monster_gold2 = 500
monster_gold3 = 1000
player_damage = 10
player_crticaldamage = player_damage * 2
monster_damage1 = 10
monster_damage2 = 5
monster_damage3 = 15
input_1 = lambda: str(random.randint(1, 9))  #1~9의 숫자를 랜덤으로 뽑음
input_2 = lambda: str(random.randint(1, 3))   #1~3의 숫자를 랜덤으로 뽑음
input_4 = lambda: str(random.randint(0, 1))   #0~1의 숫자를 랜덤으로 뽑음
player_hittime = int(input_1())
player_lasthit = int(input_4())
player_finalhittime = player_hittime + player_lasthit
monster_hittime1 = int(input_2())
monster_hittime2 = int(input_2())
monster_hittime3 = int(input_2())
monster_finalhittime = monster_hittime1 + monster_hittime2 + monster_hittime3
monster_damage = (monster_damage1 * monster_hittime1) + (monster_damage2 * monster_hittime2) + (monster_damage3 * monster_hittime3)
player_fhp = player_hp - monster_damage
monster_fhp1 = monster_hp1 - (player_damage * player_hittime) - (player_crticaldamage * player_lasthit)
monster_fhp2 = monster_hp2 - (player_damage * player_hittime) - (player_crticaldamage * player_lasthit)
monster_fhp3 = monster_hp3 - (player_damage * player_hittime) - (player_crticaldamage * player_lasthit)
monster_fhp = monster_fhp1 + monster_fhp2 + monster_fhp3
gold_final = gold + (50 * player_hittime)
monster_gold_final1 = monster_gold1 - (50 * player_hittime)
monster_gold_final2 = monster_gold2 - (50 * player_hittime)
monster_gold_final3 = monster_gold3 - (50 * player_hittime)
potion = 25
input_3 = lambda: str(random.randint(0, 1))   #0~1의 숫자를 랜덤으로 뽑음
potion_used = int(input_3())                  # 포션 사용 여부를 미리 숫자로 받아옴
player_mhp = player_fhp + (potion * potion_used)  #포션 사용후 체력
monster_gold_final = monster_gold_final1 + monster_gold_final2 + monster_gold_final3




if player_fhp <= 0:              # 체력이 0 이하이면
    print("게임 오버")
elif player_fhp < 20:            # 아니고, 20보다 적으면
    print("체력이 부족해요! 포션을 마시세요")
else:                    # 둘 다 아니면
    print("모험을 계속해요")



if monster_fhp <= 0:              # 0 "이하"이면 (<= 크거나 같다)
    print("축하해요! 승리했어요!")
elif monster_fhp < 20:            # 아니고, 20보다 적으면
    print("화이팅")
else:                        # 그렇지 않으면
    print("남은 적 총Hp", monster_fhp - player_hittime * 10, "만 더 데미지를 주면 잡을수 있어요!")



inventory = ["포션", "나무 검", "방패", "열쇠"]   # 가방
if potion_used == 1:                  # 포션을 사용했다면 (1인 경우)
  inventory.remove("포션")            # 포션을 먹어서 사용됨
  print("포션을 먹었어요!")



monsters = ["슬라임", "황금고블린", "오크"]   # 이번 웨이브의 몬스터
for m in monsters:
    print("용사가", m, "을(를) 공격했어요!")
print("웨이브 클리어!")                   # 반복이 끝난 뒤 한 번만



print(f"공격 횟수: {player_finalhittime}")
print(f"맞은 횟수: {monster_finalhittime}")
print(player_name, "남은 체력:", player_fhp, "/ 골드:", gold_final)
# monster_name 대신 monster_name1, 2, 3를 활용하여 수정
print(f"몬스터들({monster_name1}, {monster_name2}, {monster_name3}) 남은 총 체력: {monster_fhp} / 남은 총 골드: {monster_gold_final}")

import matplotlib.pyplot as plt   # 그래프 도구 불러오기
import pandas as pd                # 표 데이터 도구 불러오기
df = pd.DataFrame({"Name": ["HERO", "GOLDEN GOBLIN", "SLIME", "ORC"],
                   "New_Hp": [player_fhp, monster_fhp1, monster_fhp2, monster_fhp3]})   # HP 현황
print(df)
plt.bar(df["Name"], df["New_Hp"], color=['red', 'green', 'green', 'yellow'])
plt.title("HP")
plt.show()
```
